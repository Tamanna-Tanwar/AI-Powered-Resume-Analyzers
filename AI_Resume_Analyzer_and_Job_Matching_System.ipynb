```python
%%writefile resume_analyzer.py

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

from skills import SKILLS, SKILL_ALIASES


def extract_skills(text):
    """Find technical skills in the given text."""

    found_skills = []

    text_lower = text.lower()

    for skill in SKILLS:
        if skill.lower() in text_lower:
            found_skills.append(skill)

    # Handle common abbreviations
    for alias, full_name in SKILL_ALIASES.items():
        if alias.lower() in text_lower:
            if full_name not in found_skills:
                found_skills.append(full_name)

    return found_skills


def get_matching_skills(resume_skills, job_skills):
    """Find skills present in both resume and job description."""

    return [
        skill for skill in resume_skills
        if skill in job_skills
    ]


def get_missing_skills(resume_skills, job_skills):
    """Find job skills missing from the resume."""

    return [
        skill for skill in job_skills
        if skill not in resume_skills
    ]


def calculate_keyword_score(matching_skills, job_skills):
    """Calculate percentage based on matching skills."""

    if not job_skills:
        return 0

    return (len(matching_skills) / len(job_skills)) * 100


def calculate_nlp_score(resume_text, job_description):
    """Calculate TF-IDF cosine similarity."""

    documents = [resume_text, job_description]

    vectorizer = TfidfVectorizer(stop_words="english")

    matrix = vectorizer.fit_transform(documents)

    similarity = cosine_similarity(
        matrix[0:1],
        matrix[1:2]
    )[0][0]

    return similarity * 100


def get_recommendation(score):
    """Generate a simple recommendation."""

    if score >= 80:
        return "Strong Match"
    elif score >= 60:
        return "Good Match"
    elif score >= 40:
        return "Moderate Match"
    else:
        return "Needs Improvement"


def analyze_resume(resume_text, job_description):

    resume_skills = extract_skills(resume_text)
    job_skills = extract_skills(job_description)

    matching_skills = get_matching_skills(
        resume_skills,
        job_skills
    )

    missing_skills = get_missing_skills(
        resume_skills,
        job_skills
    )

    keyword_score = calculate_keyword_score(
        matching_skills,
        job_skills
    )

    nlp_score = calculate_nlp_score(
        resume_text,
        job_description
    )

    final_score = (
        keyword_score * 0.6
        + nlp_score * 0.4
    )

    recommendation = get_recommendation(final_score)

    return {
        "resume_skills": resume_skills,
        "job_skills": job_skills,
        "matching_skills": matching_skills,
        "missing_skills": missing_skills,
        "keyword_score": round(keyword_score, 2),
        "nlp_score": round(nlp_score, 2),
        "final_score": round(final_score, 2),
        "recommendation": recommendation
    }
```
